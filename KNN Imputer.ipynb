KNN Imputer

```python
import pandas as pd
import numpy as np
from sklearn.impute import KNNImputer

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

df=pd.read_csv(R"C:\Users\Loay\Desktop\Important Data\titanic_clean.csv")
data = df[['Age', 'Fare']]
data
```

```python
data["Age"]=data["Age"].mask(np.random.rand(len(data))<0.1)

knn_imputer=KNNImputer(n_neighbors=5)
data_knn=pd.DataFrame(knn_imputer.fit_transform(data),columns=["Age","Fare"])

print("before Imputation")
print(data.head())
print("after imputation")
print(data_knn.head())
```
